```python
import re

# regular expression
# its a pattern used to 
#   search text
# extract part of search
# valid formats
# clean or transform strings

# 1.functions
# search() - first match anywhere
# match() - match form start of string only
# findall() - all matches
# finditer() - iterator of match objects
# sub() - replace matches
# split() - split strinb by pattern
# compile() - pre compile pattern



# 2. literal characters

print(re.search('cat','my cat is here'))

# 3. special character
# \d - digit(0-9)
# \D - non digit
# \w - letter, digit, underscore
# \W - non word
# \s - whitespace
# \S - non whitespace

print(re.search(r"\D","my cat is here"))



# 4. Anchors - position matters
# ^ - start of the string 
# $ - end of the string 



# 5. quantifiers - how many times? 
# * - 0 or more
# + - 1 or more
# ? - 0 or 1
# {n} - exactly n
# {n,} - atleast n
# {n,m} - between n and m 


# 6. character sets[]
# [0-9] - digit 
# [a-z] - lowercase 
# [A-Z] - uppercase 
# [a-zA-Z] - letter 

# 7. groups() capturing data 
# groups allow you to extracct parts of the match 
# m = re.search(r'(\d{3}-(\d{4}))','555-1234')
# m.group(1) - 555 
# m.group(2) - 1234

# 8. non capturing groups(?: )
# used when you dont want extraction, only logic
# 9. alteration | - cat|dog 


# 10. flags - case insensitive,, multiline etc
# common flags 
# re.IGNORECASE - ignore case 
# re.MULTILINE - ^ & $ per line
# re,DOTALL - . matches newline 
#  re.VERBOSE - readable regex 

# 11. raw string s r""
# always use raw string for regex 

```
